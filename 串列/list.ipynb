## 串列內容的個數

len(a)取得a這個串列中的個數。
len 是 length(長度)的意思

```python
print(len(scores))
print(scores[-1])
```

```python
scores.append(78)
print(scores)
```

```python
weight=[12]*6
print(weight)
```

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
for e in scores:
    print(e)
```

```python
for i,e in enumerate(scores):
    print(i,e)
```

```python
#初始化串列，包含91個0
fib=[0]*91
print(fib)
fib[0]=1
fib[1]=1
for i in range(2,91):
    fib[i]=fib[i-1]+fib[i-2]
print(fib[90])
```
