```python
import pyautogui as pyg
pyg.sleep(2) #sleep 

while True:
    im=pyg.screenshot()
    obs=False
    for i in range(15):
        c=im.getpixel((150+i,325))
        if c ==(83,83,83):
            obs=True
            break
    if obs == True:
      pyg.press('space')
```
